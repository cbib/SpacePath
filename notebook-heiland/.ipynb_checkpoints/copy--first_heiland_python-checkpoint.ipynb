```python
# # 
def clean_matrixlike_df(df):
    df = df.sort_values(by='mz', ascending=True)
    df = df.drop_duplicates()
    return df
```

```python
def as_adata_custom(mydata1, mycoords1):
    adata = ad.AnnData(
        X=mydata1.T,
        obs=mycoords1,
        obsm={"spatial": np.array(mycoords1[["x","y"]])},
        varm={"transcripts": np.array(mydata1.index.tolist())} 
        # TODO: change to metabolites if squidpy has option to change slot name
        )
    # set pixels plot sizes
    adata.uns['spatial'] = {'my_slice_name': {
        'scalefactors': {'fiducial_diameter_fullres': 100,
                         'spot_diameter_fullres': 1,
                         'tissue_hires_scalef': 0.5,
                         'tissue_lowres_scalef': 0.5}}}
    return adata
```

```python
def open_return_adata(sample):
    mydata1 = pd.read_csv(
        os.path.join(mydata_dir, "heiland-msi", 
                 f"MALDI_{sample}.tsv"), 
        sep="\t", comment="#", header=0, index_col=0)

    mydata1 = clean_matrixlike_df(mydata1)

    mycoords1 = pd.read_csv(
        os.path.join(mydata_dir, "heiland-msi", 
                 f"coords_{sample}.tsv"), 
        sep="\t", comment="#", header=0, index_col=0)

    mycoords1 = mycoords1.set_index("barcodes")
    
    adata = as_adata_custom(mydata1, mycoords1)
    
    return adata
```

```python
def compare_do_message1(adata, me_annots):
    good = set(adata.var.index.to_numpy(dtype=float)).intersection(set(
        me_annots.annotationtbl["mz"].to_numpy()
    ))
    print("annotated mzs in original data  : ",  adata.var.shape[0])
    print(" obtained KEGG for this nb of annotations: ", len(good)) 
    # how many mz, in this sample, have KEGG annotations.
    
```

```python
def get_summary_dict(me_gmt, adata, sample):
    pathway_elems_counter_dict = dict()
    for p_id in me_gmt.gmt_dict.keys():
        pname = me_gmt.pathways_ids.loc[
            me_gmt.pathways_ids['pathway_id'] == p_id, "pathway_name"].item()
        intersec = set(adata.var.index.astype(str).tolist()).intersection(
            set(me_gmt.gmt_dict[p_id]))
        if len(intersec) > 0:
            pathway_elems_counter_dict[pname] = len(intersec)
    
    foo = pd.DataFrame.from_dict(
        pathway_elems_counter_dict, orient="index")
    
    foo = foo.reset_index()
    foo.columns = ['pathwayName', 'count']
    
    unique_counts = foo['count'].unique().tolist()
    summarydict = dict()
    for a_count in unique_counts:
        pws_here = foo.loc[foo["count"] == a_count, "pathwayName"].tolist()
        summarydict[int(a_count)] = "; ".join(pws_here) 
        
    json_out_file =  f"{sample}-summary-pws-counts.json"   
    
    with open(json_out_file, "w") as f:
        json.dump(summarydict, f)    
    return summarydict  # TODO: save json summarydict
```

```python
def summary_prep_df(summarydict):
    l1 = list()
    l2= list()
    for j in summarydict.keys():
        l1.append(j)
        l2.append(len(summarydict[j].split("; ")))
    data4plot = pd.DataFrame({"how many pathways":  l2,
                             "number of m/z features": l1})
    satisf = data4plot.loc[data4plot[
        "number of m/z features"] >= 2, "how many pathways"].sum()
    print("number of pathways with at least two features:  ", satisf)
    mycateg = [str(x) for x in sorted(
        data4plot["how many pathways"].unique().tolist(),
            reverse=True)]
    data4plot["how many pathways"] = pd.Categorical(
        data4plot["how many pathways"].astype(str), 
        categories=mycateg)
    return data4plot
```

```python
def plot_summary(data4plot, sample):
    sns.set_theme()
    sns.set(font_scale=1)
    cvals  = [1, 25, 50]
    colors = ["lightgray","salmon","red"]
    
    norm=plt.Normalize(min(cvals),max(cvals))
    tuples = list(zip(map(norm,cvals), colors))
    cmap = matplotlib.colors.LinearSegmentedColormap.from_list("reds", tuples)
    mymax = data4plot["number of m/z features"].max() * 3
    g = plt.figure(figsize=(5.4, 5))
    sns.scatterplot(data = data4plot, 
                    y="how many pathways", 
                   x="number of m/z features", 
                    color="firebrick",
                    #hue="number of m/z features",
                    #cmap=cmap,
                    edgecolor="black",
                    size="number of m/z features", 
                    sizes=(5, mymax), 
                    alpha=0.5,
                    legend=None
                   ).set_xlim(
        0.5, data4plot['number of m/z features'].max() +1)
    #plt.yticks(mycateg)    
    plt.title(sample)
    plt.savefig(f"summary-bubbles-{sample}.pdf")
    plt.close()
```

#   Sample by sample matrices

```python
samples = [
    "248_T", "259_T", "260_T", "262_T", "269_T", "275_T"
]
#samples = ["262_T"]
```

```python
adatas_hei = dict()
```

```python
#  keep each sample in a dict of adatas
for sample in samples:
    print("opening and saving in Dict[str, AnnData], the sample: ", sample)
    adi = open_return_adata(sample)
    adatas_hei[sample] = adi
    print()
```

```python
for sample in samples:
    print("processing sample: ", sample)
    adi = adatas_hei[sample]
    compare_do_message1(adi, me_annots)
    summarydict = get_summary_dict(me_gmt, adi, sample)
    data4plot = summary_prep_df(summarydict)
    plot_summary(data4plot, sample)
    print()
```
